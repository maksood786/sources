```python
from pyspark.sql.functions import avg, sum

silver_path = "/Volumes/sources_catalog/silver/cleaned/crypto"
gold_path = "/Volumes/sources_catalog/gold/business_layer/crypto"

silver_df = spark.read.format("delta").load(silver_path)

gold_df = silver_df.groupBy("symbol").agg(
    avg("current_price").alias("avg_price"),
    sum("market_cap").alias("total_market_cap"),
    sum("total_volume").alias("total_volume")
)

gold_df.write.format("delta") \
  .mode("overwrite") \
  .save(gold_path)
```
