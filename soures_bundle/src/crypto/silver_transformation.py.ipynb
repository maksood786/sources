```python
from pyspark.sql.functions import col
bronze_path = "/Volumes/sources_catalog/bronze/raw_data/crypto"
silver_path = "/Volumes/sources_catalog/silver/cleaned/crypto"



bronze_df = spark.read.format("delta").load(bronze_path)

silver_df = bronze_df.select(
    col("id").alias("coin_id"),
    col("symbol"),
    col("name"),
    col("current_price"),
    col("market_cap"),
    col("total_volume"),
    col("high_24h"),
    col("low_24h"),
    col("price_change_percentage_24h")
).dropna()

silver_df.write.format("delta") \
  .mode("overwrite") \
  .save(silver_path)
```
